```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# 1. Load datasets
PATH_ROUND1 = os.path.join("data", "round1_7755.csv")
PATH_FINAL = os.path.join("data", "final_corpus_9631.csv")
df_r1 = pd.read_csv(PATH_ROUND1)
df_final = pd.read_csv(PATH_FINAL)
print(f"Loaded Round 1 dataset:      {len(df_r1):,} samples")
print(f"Loaded Final Corpus dataset:  {len(df_final):,} samples")
# 2. Macro-label mapping
MACRO_LABEL_MAP = {
    "explicit": "Toxic Explicit",
    "explicit_mild_profanity": "Toxic Explicit",
    "sarcasm": "Toxic Implicit",
    "belittling_competence": "Toxic Implicit",
    "passive_aggressive": "Toxic Implicit",
    "implicit_threat": "Toxic Implicit",
    "contempt": "Toxic Implicit",
    "legitimate_frustration_at_product": "Non-toxic Frustration",
    "neutral_polite": "Non-toxic Civil",
    "harsh_but_civil_criticism": "Non-toxic Civil"
}
# 3. Compute Table 3 statistics
def compute_table_statistics(df_r1: pd.DataFrame, df_final: pd.DataFrame) -> pd.DataFrame:
    def get_column_stats(df: pd.DataFrame):
        total = len(df)
        passing = df[df["score"] >= 0.3]
        n_pass = len(passing)
        
        n_easy = (passing["band"] == "easy").sum()
        n_med = (passing["band"] == "medium").sum()
        n_hard = (passing["band"] == "hard").sum()
        
        macro = df["subtype"].map(MACRO_LABEL_MAP)
        n_exp = (macro == "Toxic Explicit").sum()
        n_imp = (macro == "Toxic Implicit").sum()
        n_frust = (macro == "Non-toxic Frustration").sum()
        n_civil = (macro == "Non-toxic Civil").sum()
        
        return [
            f"{total:,}",
            f"{n_pass:,} ({n_pass/total*100:.1f}%)",
            f"{n_easy:,} ({n_easy/n_pass*100:.1f}%)",
            f"{n_med:,} ({n_med/n_pass*100:.1f}%)",
            f"{n_hard:,} ({n_hard/n_pass*100:.1f}%)",
            "",
            f"{n_exp:,} ({n_exp/total*100:.1f}%)",
            f"{n_imp:,} ({n_imp/total*100:.1f}%)",
            f"{n_frust:,} ({n_frust/total*100:.1f}%)",
            f"{n_civil:,} ({n_civil/total*100:.1f}%)"
        ]
    
    statistics_index = [
        "Total samples",
        "Threshold-passing (s(x) >= 0.3)",
        "  N_Easy",
        "  N_Medium",
        "  N_Hard",
        "Macro-label distribution:",
        "  Toxic Explicit",
        "  Toxic Implicit",
        "  Non-toxic Frustration",
        "  Non-toxic Civil"
    ]
    
    col_r1 = get_column_stats(df_r1)
    col_final = get_column_stats(df_final)
    
    return pd.DataFrame({
        "Statistic": statistics_index,
        "Round 1": col_r1,
        "Round 1 + 2": col_final
    })
table3 = compute_table_statistics(df_r1, df_final)
print("\n" + "=" * 75)
print("Table 3: Effect of supplementary generation on corpus composition.")
print("Round 1 denotes initial generation; Round 1 + 2 includes supplementary generation.")
print("=" * 75)
print(table3.to_string(index=False))
print("=" * 75)
```
